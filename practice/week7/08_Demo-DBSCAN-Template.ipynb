```python
import numpy as np
import matplotlib
from matplotlib import pyplot as plt

np.set_printoptions(precision=3, suppress=True)
```

## 1. Dữ liệu giả định
(Cùng dữ liệu giả với `03_Demo-kMeans.ipynb`)

```python
np.random.seed(1)
x1,y1 = np.random.normal(loc=2,scale=1,size=50), np.random.normal(loc=5,scale=1,size=50)
x2,y2 = np.random.normal(loc=5,scale=1,size=75), np.random.normal(loc=1,scale=1,size=75)
x3,y3 = np.random.normal(loc=8,scale=1,size=100), np.random.normal(loc=7,scale=1,size=100)
```

```python
fig = plt.figure(figsize=(6,6))
plt.xlim([-2,12])
plt.ylim([-2,12])

plt.scatter(x1, y1)
plt.scatter(x2, y2)
plt.scatter(x3, y3)
plt.show()
```

### Ghép nối dữ liệu lại thành 1 tập

```python
dat1 = np.stack([x1,y1]).T
dat1.shape
```

```python
dat2 = np.stack([x2,y2]).T
dat3 = np.stack([x3,y3]).T
```

```python
dat = np.vstack([dat1, dat2, dat3])
dat.shape
```

### Sắp ngẫu nhiên các dòng

```python
np.random.seed(2)
np.random.shuffle(dat)
```

### Kiểm tra lại với biểu đồ scatter

```python
fig = plt.figure(figsize=(6,6))
plt.xlim([-2,12])
plt.ylim([-2,12])

plt.scatter(dat[:,0], dat[:,1], color='y')
plt.show()
```

## 2. Thực hiện phân cụm bằng DBSCAN

**=== Import lớp DBSCAN từ thư viện sklearn ===**

```python
from sklearn.cluster import DBSCAN
```

### 2.1. Khởi tạo model và fit vào dữ liệu

```python
model = DBSCAN(eps=1, min_samples=5)
```

**=== Thực hiện fit vào dữ liệu ===**

```python
model.fit(dat)
```

### 2.2. Kiểm tra kết quả các cluster tìm được

**=== Lấy ra nhãn các clusters ===**

```python
cluster = model.labels_
```

```python
cluster
```

**=== Đếm xem mỗi cluster có bao nhiêu điểm ===**

```python
print(np.unique(cluster, return_counts=True))
```

**=== Hiển thị lên biểu đồ ===**

```python
for label in np.unique(cluster):
    points = dat[cluster == label]
    if label == -1:
        plt.scatter(points[:, 0], points[:, 1], c='black', marker='x', label='noise')
    else:
        plt.scatter(points[:, 0], points[:, 1], label=f'cluster {label}')
plt.legend()
plt.grid(True)
```

### 2.3. Tính hiệu suất phân cụm

**=== Thử tính với hàm built-in của Sklearn ===**

```python
from sklearn.metrics import silhouette_score
```

```python
n_clusters = len(set(cluster) - {-1})
if n_clusters >= 2:
    print('Silhouette, giu noise:', silhouette_score(dat, cluster))
else:
    print('Khong du 2 cum de tinh Silhouette khi giu noise.')
```

**=== Tính lại với các cụm không chứa điểm noise ===**

```python
import pandas as pd
```

```python
mask = cluster != -1
dat2 = dat[mask]
cluster2 = cluster[mask]
print('So mau sau khi bo noise:', len(dat2))
```

```python
s_list = []

for i, sample in enumerate(dat2):
    label = cluster2[i]
    other_mask = np.arange(len(dat2)) != i
    other_samples = dat2[other_mask]
    other_labels = cluster2[other_mask]
    distances = np.sqrt(np.sum((other_samples - sample) ** 2, axis=1))
    distance_frame = pd.DataFrame({'distance': distances, 'cluster': other_labels})
    mean_distances = distance_frame.groupby('cluster')['distance'].mean()
    a = mean_distances.loc[label] if label in mean_distances.index else 0.0
    other_cluster_means = mean_distances.drop(index=label, errors='ignore')
    b = other_cluster_means.min() if not other_cluster_means.empty else 0.0
    s_list.append((b - a) / max(a, b) if max(a, b) > 0 else 0.0)
```

```python
if len(np.unique(cluster2)) >= 2:
    print('Silhouette trung binh sau khi bo noise:', np.mean(s_list))
else:
    print('Khong du 2 cum sau khi bo noise de tinh Silhouette.')
```
