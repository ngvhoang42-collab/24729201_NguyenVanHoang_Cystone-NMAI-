**Question 40.** What changes when kNN K increases from 1 to a large value? What changes when k-Means K increases from 2 to 10? Explain why these are different effects.

```python
cau_40 = """
kNN: larger K votes across more neighbors, smoothing predictions toward the overall common class; too-large K can erase local structure. k-Means: larger K creates more, smaller and more detailed clusters, and inertia cannot increase (it usually falls). This changes clustering resolution, not the number of neighbors voting on a class.
"""
```
