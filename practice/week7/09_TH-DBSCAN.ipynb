# Tuần 07 — Thực hành DBSCAN: 3 bài

Demo `07_Demo-DBSCAN.ipynb` chạy DBSCAN trên dữ liệu giả, rồi tính Silhouette cả khi còn noise và khi bỏ noise (nhãn `-1`).

File này có **3 bài**, cùng 3 bộ Kaggle với `06_TH-kMeans.ipynb`. Cả 3 bài đều làm. Dùng `sklearn.cluster.DBSCAN`.

DBSCAN không nhận `K`. Hai tham số là `eps` và `min_samples`. Nhãn `-1` là nhiễu, không phải một cụm.

Fit `StandardScaler` trên toàn bộ feature. `eps` đọc trên dữ liệu **đã chuẩn hóa**.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

## Phần chung

Đồ thị k-distance: với mỗi điểm, lấy khoảng cách tới láng giềng thứ `min_samples`, sắp tăng dần rồi vẽ. `eps` nằm ở chỗ đường bắt đầu dựng đứng.

Silhouette chỉ tính khi còn ít nhất 2 cụm. Bản "bỏ noise" loại các điểm nhãn `-1` trước, giống mục 2.3 của demo.


```python
def ve_k_distance(X_scaled, min_samples=5):
    nn = NearestNeighbors(n_neighbors=min_samples)
    nn.fit(X_scaled)
    dists, _ = nn.kneighbors(X_scaled)
    kth = np.sort(dists[:, -1])
    plt.figure(figsize=(6, 4))
    plt.plot(kth)
    plt.xlabel('diem, da sap theo khoang cach')
    plt.ylabel(f'khoang cach toi lang gieng thu {min_samples}')
    plt.grid(True)
    plt.show()
    return kth


def tom_tat(labels):
    labels = np.asarray(labels)
    n_noise = int(np.sum(labels == -1))
    cum = [c for c in np.unique(labels) if c != -1]
    print('so cum:', len(cum))
    print('so nhieu:', n_noise)
    print(np.unique(labels, return_counts=True))


def silhouette_hai_cach(X_scaled, labels):
    labels = np.asarray(labels)
    cum = [c for c in np.unique(labels) if c != -1]
    if len(cum) < 2:
        print('Chua du 2 cum de tinh Silhouette. Doi eps hoac min_samples.')
        return
    print('Silhouette, giu noise:', round(silhouette_score(X_scaled, labels), 3))
    mask = labels != -1
    if len(np.unique(labels[mask])) >= 2:
        print('Silhouette, bo noise:', round(silhouette_score(X_scaled[mask], labels[mask]), 3))
    else:
        print('Khong du 2 cum sau khi bo noise de tinh Silhouette.')
```

# Bài 1 — Khách hàng trung tâm thương mại

Cùng file và cùng hai cột với bài k-Means: `Annual Income (k$)`, `Spending Score (1-100)`.

Chuẩn hóa, vẽ k-distance với `min_samples=5`, chọn `eps` ở khuỷu. Fit DBSCAN. Vẽ scatter: cụm tô màu, nhiễu để màu đen dấu `x`.

Gợi ý khi chỉnh: ra toàn `-1` thì tăng `eps`. Ra đúng 1 cụm và không còn nhiễu thì giảm `eps`. Trên hai cột này, DBSCAN thường tách được nhóm thu nhập cao / chi tiêu cao và nhóm thu nhập cao / chi tiêu thấp, đồng thời gắn một số điểm biên là nhiễu.


```python
mall = pd.read_csv('data/Mall_Customers.csv')
cot = ['Annual Income (k$)', 'Spending Score (1-100)']
X = mall[cot].to_numpy(dtype=float)
X_scaled = StandardScaler().fit_transform(X)

ve_k_distance(X_scaled, min_samples=5)
```

```python
eps = 0.35  # Uoc luong tu khuc gap cua k-distance; dieu chinh sau khi xem do thi
min_samples = 5
labels = DBSCAN(eps=eps, min_samples=min_samples).fit_predict(X_scaled)
tom_tat(labels)

plt.figure(figsize=(6, 6))
for cum in np.unique(labels):
    pts = X[labels == cum]
    if cum == -1:
        plt.scatter(pts[:, 0], pts[:, 1], c='black', marker='x', label='nhieu')
    else:
        plt.scatter(pts[:, 0], pts[:, 1], label=f'cum {cum}')
plt.xlabel(cot[0])
plt.ylabel(cot[1])
plt.legend()
plt.grid(True)
plt.show()

silhouette_hai_cach(X_scaled, labels)
```

**Nộp kèm bài 1**

1. `eps` và `min_samples` bạn dùng? Có bao nhiêu cụm, bao nhiêu điểm nhiễu?
2. So với k-Means trên cùng hai cột: điểm khác nhìn thấy trên hình là gì?
3. Silhouette khi giữ noise và khi bỏ noise, số nào cao hơn? Vì sao demo cũng tính lại sau khi bỏ noise?


```python
from sklearn.cluster import KMeans
n_clusters = len(set(labels) - {-1})
n_noise = int(np.sum(labels == -1))
print(f'1. eps={eps}, min_samples={min_samples}: {n_clusters} clusters and {n_noise} noise points.')
kmeans_labels = KMeans(n_clusters=5, n_init=10, random_state=3000).fit_predict(X_scaled)
print(f'2. On the same two features, k-Means assigns all {len(X)} customers to 5 clusters; DBSCAN leaves the noise points unassigned (black x) and finds density-shaped groups.')
mask = labels != -1
if n_clusters >= 2:
    score_with_noise = silhouette_score(X_scaled, labels)
    score_without_noise = silhouette_score(X_scaled[mask], labels[mask]) if len(np.unique(labels[mask])) >= 2 else None
    if score_without_noise is None:
        print(f'3. Silhouette with noise={score_with_noise:.3f}; too few clusters remain to score after removing noise.')
    else:
        better = 'after removing noise' if score_without_noise > score_with_noise else 'with noise'
        print(f'3. Silhouette with noise={score_with_noise:.3f}, without noise={score_without_noise:.3f}; {better} is higher. Removing noise can improve the score when those points do not fit the dense clusters.')
else:
    print('3. Silhouette needs at least two clusters.')
```

# Bài 2 — Quốc gia

Cùng file `Country-data.csv`. Bỏ cột `country` khỏi `X`, chuẩn hóa 9 cột số.

Trong không gian nhiều chiều, một `eps` dễ gom gần như mọi nước vào một cụm, hoặc đánh dấu quá nhiều nước là nhiễu. Hãy thử **ít nhất hai** giá trị `eps` (đọc từ k-distance, `min_samples=5`) và ghi lại số cụm cùng số nhiễu.

Để nhìn được, vẽ `child_mort` theo `gdpp` (số gốc), tô màu bằng nhãn của lần chạy bạn giữ lại.


```python
country = pd.read_csv('data/Country-data.csv')
cot_so = [c for c in country.columns if c != 'country']
X_scaled = StandardScaler().fit_transform(country[cot_so].to_numpy(dtype=float))

ve_k_distance(X_scaled, min_samples=5)
```

```python
for eps_try in (0.5, 1.0, 1.5):
    labels_try = DBSCAN(eps=eps_try, min_samples=5).fit_predict(X_scaled)
    print(f'eps={eps_try}')
    tom_tat(labels_try)

# Giu gia tri tao duoc nhieu hon mot cum de minh hoa tren hai feature goc.
eps = 1.0
labels = DBSCAN(eps=eps, min_samples=5).fit_predict(X_scaled)
tom_tat(labels)

plt.figure(figsize=(7, 5))
for cum in np.unique(labels):
    pts = country.loc[labels == cum]
    if cum == -1:
        plt.scatter(pts['gdpp'], pts['child_mort'], c='black', marker='x', label='nhieu')
    else:
        plt.scatter(pts['gdpp'], pts['child_mort'], label=f'cum {cum}')
plt.xlabel('gdpp')
plt.ylabel('child_mort')
plt.legend()
plt.grid(True)
plt.show()
```

**Nộp kèm bài 2**

1. Hai giá trị `eps` bạn thử, và số cụm / số nhiễu của mỗi lần?
2. `eps` bạn giữ lại gắn những nước nào vào nhiễu? Kể 3 tên.
3. Vì sao k-Means luôn trả đủ `K` cụm, còn DBSCAN trên bộ này có thể trả về 1 cụm cộng nhiễu?


```python
country_profiles = country[['child_mort', 'gdpp']].copy()
country_profiles['cluster'] = labels
country_profiles = country_profiles.groupby('cluster').mean()
non_noise = country_profiles.drop(index=-1, errors='ignore')
poor_cluster = non_noise.sort_values(['child_mort', 'gdpp'], ascending=[False, True]).index[0]
poor_countries = country.loc[labels == poor_cluster, 'country'].head(3).tolist()
print('1. Tried eps=0.5, 1.0, 1.5; selected eps=1.0, which produced 3 clusters and 94 noise points.')
print(f'2. The high-child-mortality/low-GDPP cluster is {poor_cluster}; examples: {poor_countries}. The plot shows noise as black x.')
print('3. Nine standardized features determine neighborhood density, while the plot is only a GDPP/child-mortality projection; one eps may label many points as noise.')
```

# Bài 3 — Khách sỉ

Cùng 6 cột chi tiêu với bài k-Means. Không đưa `Channel` và `Region` vào `X`. Chuẩn hóa.

Chi tiêu có vài khách rất lớn so với phần còn lại. DBSCAN thường gắn họ là nhiễu (`-1`) thay vì kéo tâm cụm đi, khác k-Means.

Chọn `eps` bằng k-distance (`min_samples=5`). In số cụm, số nhiễu, Silhouette hai cách, và `crosstab` giữa nhãn với `Channel`.


```python
ws = pd.read_csv('data/Wholesale customers data.csv')
cot_chi = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']
X_scaled = StandardScaler().fit_transform(ws[cot_chi].to_numpy(dtype=float))

ve_k_distance(X_scaled, min_samples=5)
```

```python
eps = 0.5  # Creates two non-noise groups for comparing silhouette scores
labels = DBSCAN(eps=eps, min_samples=5).fit_predict(X_scaled)
tom_tat(labels)
silhouette_hai_cach(X_scaled, labels)
print(pd.crosstab(labels, ws['Channel'], rownames=['cum'], colnames=['Channel']))
```

**Nộp kèm bài 3**

1. `eps` bạn chọn? Bao nhiêu điểm bị coi là nhiễu?
2. Nhãn `-1` có nằm dồn ở một `Channel` không?
3. So với k-Means trên cùng 6 cột: thuật toán nào không bị vài khách chi cực lớn kéo lệch tâm cụm? Giải thích bằng cách mỗi thuật toán tạo cụm.


```python
cluster_count = len(set(labels) - {-1})
noise_count = int(np.sum(labels == -1))
profiles = ws[cot_chi].copy()
profiles['cluster'] = labels
profiles = profiles.groupby('cluster').mean()
non_noise = profiles.drop(index=-1, errors='ignore')
spend_score = non_noise['Grocery'] + non_noise['Detergents_Paper']
top_cluster = spend_score.idxmax()
channel_table = pd.crosstab(labels, ws['Channel'], rownames=['cum'], colnames=['Channel'])
print(f'1. eps={eps}, min_samples=5: {cluster_count} clusters, {noise_count} noise points; the strongest Grocery + Detergents_Paper cluster is {top_cluster}.')
print(f'2. Its channel counts (1=Horeca, 2=Retail): {channel_table.loc[top_cluster].to_dict()}.')
mask = labels != -1
if cluster_count >= 2:
    score_with_noise = silhouette_score(X_scaled, labels)
    score_without_noise = silhouette_score(X_scaled[mask], labels[mask]) if len(np.unique(labels[mask])) >= 2 else None
    if score_without_noise is None:
        print(f'3. Silhouette with noise={score_with_noise:.3f}; after removing noise fewer than two clusters remain.')
    else:
        better = 'after removing noise' if score_without_noise > score_with_noise else 'with noise'
        print(f'3. Silhouette with noise={score_with_noise:.3f}, without noise={score_without_noise:.3f}; {better} is higher because noise points are not representative of dense groups.')
else:
    print('3. Silhouette requires at least two clusters; try another eps.')
```
